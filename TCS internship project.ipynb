```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("Online Retail Data Set.csv")
```

```python
df
```

```python
df = df[(df['Quantity'] >= 1) & (df['UnitPrice'] > 0)]
```

```python
df
```

```python
df = df.dropna(subset=['CustomerID'])
```

```python
df
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
```

```python
df['Revenue'] 
```

```python
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['MonthName'] = df['InvoiceDate'].dt.strftime('%b')

```

```python
df['InvoiceDate']
```

```python
df['Year']
```

```python
df['Month']
```

```python
df['MonthName']
```

```python
q1 = df[df['Year'] == 2011].groupby(['Month','MonthName'])['Revenue'].sum().reset_index().sort_values('Month')
q1
```

```python
q2 = df[df['Country'] != 'United Kingdom'].groupby('Country').agg({'Revenue':'sum','Quantity':'sum'}).reset_index()
q2 = q2.sort_values('Revenue', ascending=False).head(10)
```

```python
q2
```

```python
q3 = df.groupby('CustomerID')['Revenue'].sum().reset_index().sort_values('Revenue', ascending=False).head(10)
q3
```

```python
q4 = df[df['Country'] != 'United Kingdom'].groupby('Country')['Quantity'].sum().reset_index()
q4
```
